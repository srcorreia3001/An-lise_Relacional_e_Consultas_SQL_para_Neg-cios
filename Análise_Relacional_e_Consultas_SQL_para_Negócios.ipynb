```python
import os
import pandas as pd
import sqlite3

#Carregando os arquivos da tabela
caminho_arquivo = '/kaggle/input/notebooks/fareselgohary003/superstore-dataset/final_dataset.xlsx'
df = pd.read_excel(caminho_arquivo)

#Criando uma conexão com o banco de dados SQlite local
conexao = sqlite3.connect('banco_vendas.db')
#Transformando o dataframe do pandas em uma tabela sql chamada vendas
df.to_sql('vendas',conexao,if_exists='replace',index=False)

print("Banco de dados SQlite criado e tabela 'vendas' pronta para uso!")

```

```python
#Verificando os nomes exatos das colunas criadas na tabela SQl
colunas_sql = pd.read_sql("PRAGMA table_info(vendas);",conexao)
display(colunas_sql)
```

```python
import pandas as pd

#Consulta ao banco de dados sql
query_basica = """
    SELECT [Customer Name], City, Sales, Profit
    FROM vendas
    WHERE Sales > 1000
    ORDER BY Sales DESC
    LIMIT 5;
"""
#Executando a consulta ao banco de dados sql e trazendo o resultado para o pandas
resultado_sql = pd.read_sql(query_basica,conexao)

#Exibindo resultado em formato de tabela
display(resultado_sql)
```

```python
import pandas as pd

#Consulta sql utilizando GROUP BY e função de agregação SUM()
query_agrupada = """
    SELECT Category, SUM(Sales) AS Faturamento_Total, SUM(Profit) AS Lucro_Total
    FROM vendas
    GROUP BY Category
    ORDER BY Faturamento_Total DESC;
"""
#Executando a consulta no banco de dados sql
resultado_grupo = pd.read_sql(query_agrupada,conexao)

#Exibindo o Resulta do Agrupado
display(resultado_grupo)
```
